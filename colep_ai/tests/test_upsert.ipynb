```python
from colep_ai.database.azure_search_client import get_search_client

client = get_search_client()

# get all unique source_files first
results = client.search(search_text="*", select=["source_file"], top=1000)
source_files = set(r["source_file"] for r in results)

for sf in sorted(source_files):
    count_result = client.search(
        search_text="*",
        filter=f"source_file eq '{sf}'",
        top=0,
        include_total_count=True
    )
    print(f"{sf}: {count_result.get_count()}")
```
