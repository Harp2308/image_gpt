```python
def extract_line_number(stem: str) -> list[int] | None:
    # Strip doc code prefix: Q01_L132_3_ or O01_O119_1_
    stripped = re.sub(r'^[A-Z]\d+_[A-Z]\d+_\d+_', '', stem, flags=re.IGNORECASE)

    # Match "Linhas_28_34_e_95" OR "L19_e_83" OR "Linha_19" OR "L13"
    match = re.search(r'(?:Linhas?|L)[_ ]*((?:\d+[_ e,]*)+)', stripped, re.IGNORECASE)
    if not match:
        return None

    return [int(n) for n in re.findall(r'\d+', match.group(1))]
```

```python
cases = [
    "Q01_M156_2_Mapa_de_controlo_da_Estampagem_Linhas_28_34_e_95",
    "O01_O119_1_Localizacao_dos_pontos_de_lubrificacao_e_processo_de_limpeza_L13",
    "O01_I054_1_Instrucao_de_Operacao_Ligar_e_desligar_Paletizador_Llinha_19_Assembly",
    "Q01_L132_3_Plano_de_Inspecao_e_Ensaio_L19_e_83_Estampagem",
]

for c in cases:
    print(extract_line_number(c))

# [28, 34, 95]
# [13]
# [19]
# [19, 83]   ← L132 is doc code, L19 is the line ref
```
